```python
from ultralytics import YOLO

# Load model (pretrained)
model = YOLO("yolov8m.pt")   # n = nano (fast), can use s/m/l/x

# Train
results = model.train(
    data="/content/baby-things-1/data.yaml",
    epochs=75,
    imgsz=640,
    batch=16,
    name="baby",
    optimizer="Adam",
    lr0=0.001,
    patience=10,
    device=0   # GPU (use 'cpu' if no GPU)
)

# Validate
metrics = model.val()

```

```python
import requests

# Replace this URL with the actual best.pt file link
FILE_URL = "/content/runs/detect/baby/weights/best.pt"
SAVE_PATH = "/content/best.pt"  # Colab's working directory

response = requests.get(FILE_URL, stream=True)
response.raise_for_status()  # check for errors

with open(SAVE_PATH, "wb") as f:
    for chunk in response.iter_content(chunk_size=8192):
        if chunk:
            f.write(chunk)

print(f"✅ best.pt downloaded to {SAVE_PATH}")
```
