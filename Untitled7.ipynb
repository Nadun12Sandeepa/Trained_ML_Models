```python
import cv2
import numpy as np
from ultralytics import YOLO
from google.colab.patches import cv2_imshow

# Load trained model
model = YOLO("/content/runs/detect/vegetable_detector/weights/best.pt")

# ----------------------------
# Preprocessing function
# ----------------------------
def preprocess_image(img_path):
    img = cv2.imread(img_path)

    # 1. Auto-Orient (OpenCV reads correctly usually, but keep safe)
    img = cv2.rotate(img, cv2.ROTATE_180) if img is None else img

    # 2. Resize (Stretch to 640x640)
    img = cv2.resize(img, (640, 640))

    # 3. Auto-Adjust Contrast (CLAHE)
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    l, a, b = cv2.split(lab)

    clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
    l = clahe.apply(l)

    lab = cv2.merge((l, a, b))
    img = cv2.cvtColor(lab, cv2.COLOR_LAB2BGR)

    # 4. Grayscale → convert back to 3 channel (VERY IMPORTANT)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    img = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

    return img


# ----------------------------
# Prediction function
# ----------------------------
def predict(image_path):
    img = preprocess_image(image_path)

    results = model(img)

    for r in results:
        boxes = r.boxes
        if boxes is not None:
            for box in boxes:
                cls = int(box.cls[0])
                conf = float(box.conf[0])
                x1, y1, x2, y2 = map(int, box.xyxy[0])

                label = model.names[cls]

                print(f"{label} ({conf:.2f})")

                # Draw bbox
                cv2.rectangle(img, (x1, y1), (x2, y2), (0,255,0), 2)
                cv2.putText(img, f"{label} {conf:.2f}",
                            (x1, y1-10),
                            cv2.FONT_HERSHEY_SIMPLEX,
                            0.6, (0,255,0), 2)

    cv2_imshow(img)


# ----------------------------
# Run
# ----------------------------
predict("/content/1.jpg")
```
